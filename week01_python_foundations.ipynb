```python
#create a variable of a bodaboda rider's day
trips = 18
avg_fare = 220.0
fuel_cost = 1100.0
rider_name = "Kelvin"

#Calculate gross income,net income, and mrgin_pct
gross_income = trips * avg_fare
net_income = gross_income - fuel_cost
margin_pct = net_income / gross_income * 100

# use type() to  confirm the data type of each variable
print(f"The data type of  variable trips is {type(trips)}")
print(f"The data type of  variable avg_fare is {type(avg_fare)}")
print(f"The data type of  variable fuel_cost is {type(fuel_cost)}")
print(f"The data type of  variable rider_name is {type(rider_name)}")
print(f"The data type of  variable gross_income is {type(gross_income)}")
print(f"The data type of  variable net_income is {type(net_income)}")
print(f"The data type of  variable margin_pct is {type(margin_pct)}")

#one line summary using an f-string formatting money with :,.2f and the margin with :.1f
print(f"{rider_name}: gross KES{gross_income:,.2f}, "
f"net KES{net_income:,.2f}, margin {margin_pct:.1f}%")

```

```python
#Lab1.3 Inventory Alerts with dictionaries and loops
inventory = [
    {'name':'rice','stock':7,'reorder_level': 13},
    {'name':'sugar','stock':21,'reorder_level': 12},
    {'name':'salt','stock':14,'reorder_level': 13},
    {'name':'Unga','stock':10,'reorder_level': 9}]

#Loop through the list and for each product
def restock_report(items):
  '''Print stock status and return a list of products to reorder'''
  to_reorder = []
  for item in items:
    if item["stock"]<=item["reorder_level"]:
      status ="REORDER"
      to_reorder.append(item["name"])
    else:
      status = "OK"
    print(f"{item['name']:<6}stock ={item['stock']:>3} {status}")
  return to_reorder

print(restock_report(inventory))
```
